### CRISP-DM Phase 4.1 - Modeling : Majority-class baseline classification

```python
import pandas as pd
import ast
from collections import Counter
from codecarbon import EmissionsTracker
```

```python
# Load the dataset
law = pd.read_csv('data/law.csv')
benchmark = pd.read_csv('data/benchmark.csv')

# Fix the format of the hazard column 
benchmark['Hazard'] = benchmark['Hazard'].apply(lambda x: x if isinstance(x, list) else ast.literal_eval(x))
```

```python
## Retrieve most frequent labels
all_labels = [label for labels in benchmark['Hazard'] for label in labels if label.strip() != '']

# Top 1 
top_1_hazard = Counter(all_labels).most_common(1)[0][0]
print(f"Most frequent label: {top_1_hazard}")

# Top 3 
top_3_hazards = [Counter(all_labels).most_common(3)[i-1][0] for i in range(1, 4)]
print(f"Top 3 most frequent labels: {top_3_hazards}")
```

```python
## Label benchmark documents with most frequent labels
tracker = EmissionsTracker()
tracker.start()

benchmark['Hazard_top1_baseline'] = benchmark['Summary'].apply(lambda x: [top_1_hazard])
benchmark['Hazard_top3_baseline'] = benchmark['Summary'].apply(lambda x: top_3_hazards)

emissions = tracker.stop()
print(f"Emissions: {emissions} kg CO₂")
```

```python
## Export .csv
benchmark.to_csv('data/benchmark_baseline.csv', index=False)
```
