### CRISP-DM Phase 4.1 - Modeling : Rule-based NLP classification

```python
import pandas as pd
import ast
from codecarbon import EmissionsTracker
```

```python
# Load the dataset
law = pd.read_csv('data/law.csv')
benchmark = pd.read_csv('data/benchmark.csv')

# Fix the format of the hazard column 
benchmark['Hazard'] = benchmark['Hazard'].apply(lambda x: x if isinstance(x, list) else ast.literal_eval(x))
```

```python
glossary = {
    # Flood
    'flood':
    ['flood', 'floods', 'flooding', 'deluge', 'downpour', 'surge', 'torrent', 'inundation', 'overflow',
     'stormwater', 'waterlogging'],
    # Drought
    'drought': 
    ['drought', 'droughts', 'arid', 'aridity', 'dry', 'desiccation', 'desiccate'],
    # Temperature extremes
    'temperature_extremes': 
    ['heat', 'heatwave', 'heatwaves', 'temperature', 'cold', 'warm', 'thermal stress', 'cooling'],
    # Sea level rise (and coastal hazards)
    'sea_level_rise':       
    ['sea level', 'coastal', 'water level', 'coast', 'tide', 'saltwater intrusion'],
    # Storm
    'storm':           
    ['storm', 'storms', 'windstorm', 'windstorms', 'wind', 'winds', 'cyclone', 'cyclones', 'hurricane', 
     'hurricanes', 'typhoon', 'typhoons', 'tornado', 'gale', 'gust', 'blizzard', 'monsoon', 'squall',
     'tempest'],
    # Wildfire
    'wildfire': 
    ['wildfire', 'wildfires', 'fire', 'fires', 'bushfire', 'combustion', 'burning'],
    # Snowmelt
    'melting':      
    ['snow', 'melt', 'melting', 'ice', 'glacier', 'permafrost', 'thawing'],
    # Erosion
    'erosion':              
    ['erosion', 'corrosion', 'desintegration', 'degradation', 'sediment loss'],
    # Other
    'other':                
    ['tsunami', 'tsunamis', 'earthquake', 'earthquakes', 'precipitation', 'soil quality', 'air quality', 
     'surface water', 'groundwater', 'desertification', 'biodiversity', 'acidification', 'tide', 'tides', 
     'hail', 'frost', 'epidemic', 'epidemics', 'disease', 'diseases', 'eruption', 'eruptions', 'lightning', 
     'avalanche', 'avalanches', 'landslide', 'landslides', 'mudslide', 'mudslides', 'disaster', 'disasters'],
}
```

```python
def rule_based_classifier(text):
    if pd.isna(text) or text == '':
        return ['none']
    labels = []
    for hazard, keywords in glossary.items():
        if any(keyword in text.lower() for keyword in keywords):
            labels.append(hazard)
    return sorted(labels) if labels else ['none']
```

```python
## Label benchmark documents with rule-based classifier
tracker = EmissionsTracker()
tracker.start()

benchmark['Hazard_rulebased'] = benchmark['Summary'].apply(rule_based_classifier)

emissions = tracker.stop()
print(f"Emissions: {emissions} kg CO₂")
```

```python
## Export .csv
benchmark.to_csv('data/benchmark_rulebased.csv', index=False)
```
