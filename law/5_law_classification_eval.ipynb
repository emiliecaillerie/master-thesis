### CRISP-DM Phase 5.1 - Evaluation : Classification models

```python
import pandas as pd
import ast
from sklearn.preprocessing import MultiLabelBinarizer
from sklearn.metrics import classification_report, hamming_loss, jaccard_score
```

```python
# Load the dataset
benchmark_baseline = pd.read_csv('data/benchmark_baseline.csv')
benchmark_rulebased = pd.read_csv('data/benchmark_rulebased.csv')
benchmark_llm = pd.read_csv('data/benchmark_llm.csv')
benchmark_llm_bis = pd.read_csv('data/benchmark_llm_bis.csv')
benchmark_llm_ter = pd.read_csv('data/benchmark_llm_ter.csv')
```

```python
models_dict = {'baseline_top1': 'top1_baseline', 'baseline_top3': 'top3_baseline',
               'rulebased': 'rulebased',
               'llm_gpt20': 'gpt20', 'llm_gpt120': 'gpt120',
               'llm_gemma4': 'gemma4', 'llm_gemma27': 'gemma27',
               'llm_minimax': 'minimax', 'llm_minimax2': 'minimax2',
               'llm_qwen480': 'qwen480', 'llm_qwen80': 'qwen80'}

# Fix the format of the hazard column
benchmark_baseline['Hazard'] = benchmark_baseline['Hazard'].apply(lambda x: x if isinstance(x, list) else ast.literal_eval(x))
for name, model in models_dict.items():
    if 'baseline' in name:
        benchmark_baseline[f'Hazard_{model}'] = benchmark_baseline[f'Hazard_{model}'].apply(lambda x: x if isinstance(x, list) else ast.literal_eval(x))
    elif 'rulebased' in name:
        benchmark_rulebased[f'Hazard_{model}'] = benchmark_rulebased[f'Hazard_{model}'].apply(lambda x: x if isinstance(x, list) else ast.literal_eval(x))
    else:
        benchmark_llm[f'Hazard_{model}'] = benchmark_llm[f'Hazard_{model}'].apply(lambda x: x if isinstance(x, list) else ast.literal_eval(x))
```

```python
HAZARDS = ['flood', 'drought', 'temperature_extremes', 'sea_level_rise', 'storm', 'wildfire',
          'melting', 'erosion', 'other', 'none']

# Binarize to handle multi-labelling
mlb = MultiLabelBinarizer()
mlb.fit([HAZARDS]) 

y = mlb.transform(benchmark_baseline['Hazard'])
y_predictions = {}
for name, model in models_dict.items():
    if 'baseline' in name:
        y_predictions[name] = mlb.transform(benchmark_baseline[f'Hazard_{model}'])
    elif 'rulebased' in name:
        y_predictions[name] = mlb.transform(benchmark_rulebased[f'Hazard_{model}'])
    else:
        y_predictions[name] = mlb.transform(benchmark_llm[f'Hazard_{model}'])
```

```python
# Precision, recall, F1-score
for name, y_pred in y_predictions.items():
    print(f"{name}:")
    print(classification_report(y, y_pred, output_dict=False, target_names=mlb.classes_, zero_division=0))
```

```python
# Extra metrics : Hamming loss and Jaccard score
for name, y_pred in y_predictions.items():
    print(f"{name}: Hamming loss = {hamming_loss(y, y_pred)}, Jaccard score = {jaccard_score(y, y_pred, average='macro', zero_division=0)}")
```
